    <br>
  <h4 style="margin-top: 0; margin-bottom: 15px; color: #2c3e50; font-size: 16px;">🔎 太字と斜体の使い分け</h4>
  <p style="margin: 0; line-height: 2.5; font-size: 14px; color: #2c3e50; text-indent: 2em;">
    太字や斜体は<b>文章中の特定の語句やフレーズを強調</b>する目的で使用される．
  </p>
  <ul style="margin-top: 1em; line-height: 2; font-size: 14px; color: #2c3e50;">
      ✅ <b>太字（strong emphasis）</b><br>
  <p style="margin: 0; line-height: 2.5; font-size: 14px; color: #2c3e50; text-indent: 1.5em;">
      太字は，読者の注意を強く引きたい場合に使用する．
  </p>
  <p style="margin: 0; line-height: 2.5; font-size: 14px; color: #2c3e50; text-indent: 1.5em;">
      ex）重要な警告や注意事項，見出し，キーワードや専門用語の定義
  </p><br>
      ✅ <i>斜体（emphasis）</i><br>
  <p style="margin: 0; line-height: 2.5; font-size: 14px; color: #2c3e50; text-indent: 1.5em;">
      斜体は，通常の文章とは異なる性質のテキストを示すのに使用する．
    </p>
  <p style="margin: 0; line-height: 2.5; font-size: 14px; color: #2c3e50; text-indent: 1.5em;">
      ex）分類学上の名称，技術用語，他言語からきた慣用句，字訳，思考，西洋文中の船名
    </p> 
  </ul>
    <p style="margin: 1.5em auto 0 auto;font-size: 13.5px;color: #444;background-color: #ffffff;padding: 16px 20px;width: 640px;line-height: 2.2;border-radius: 4px;">
      ※テキストを<b><i>太字かつ斜体</i></b>にするには，<code>***</code> と <code>___</code> のようにアスタリスク（*）または<br>
      <span style="text-indent: 1em; display: inline-block;">アンダースコア（_）を3つずつ使用する. <br><span style="text-indent: 1em; display: inline-block;">この形式は，<span style="background-color: #fffeafee;">特に重要な語句やフレーズを際立たせたい</span>場合に有効である．</span>
    </p>

​CommonMark仕様の「2.1 Characters and lines」セクションでは、Markdown文書の基本的な構成要素である「文字」と「行」について定義されています。以下にその内容を日本語で要約します。​

🔹 文字と行の定義
文字（Character）: Unicodeのコードポイントを指し、結合文字など直感的に文字と感じられないものも含まれます。​仕様上、すべてのコードポイントが文字として扱われます。​

行（Line）: 改行（U+000A）や復帰（U+000D）を含まない文字の並びで、行末記号またはファイルの終わりで終わります。​

行末記号（Line ending）: 以下のいずれかです。

改行（U+000A）

復帰（U+000D）で、次に改行が続かないもの

復帰とそれに続く改行の組み合わせ​

空行（Blank line）: 文字を含まない行、またはスペース（U+0020）やタブ（U+0009）のみを含む行です。​

🔹 文字クラスの定義
空白文字（Whitespace character）: スペース（U+0020）、タブ（U+0009）、改行（U+000A）、行タブ（U+000B）、フォームフィード（U+000C）、復帰（U+000D）を含みます。​

空白（Whitespace）: 1つ以上の空白文字の連続です。​

Unicode空白文字（Unicode whitespace character）: UnicodeのZsカテゴリに属するコードポイント、またはタブ（U+0009）、復帰（U+000D）、改行（U+000A）、フォームフィード（U+000C）です。​

Unicode空白（Unicode whitespace）: 1つ以上のUnicode空白文字の連続です。​

スペース（Space）: U+0020のコードポイントです。​

非空白文字（Non-whitespace character）: 空白文字以外のすべての文字です。​

ASCII句読点文字（ASCII punctuation character）: 以下の記号を含みます。

!, ", #, $, %, &, ', (, ), *, +, ,, -, ., /（U+0021–2F）

:, ;, <, =, >, ?, @（U+003A–0040）

[, \, ], ^, _, `（U+005B–0060）

{, |, }, ~（U+007B–007E）​

句読点文字（Punctuation character）: ASCII句読点文字、またはUnicodeのカテゴリPc、Pd、Pe、Pf、Pi、Po、Psに属する文字です。

このセクションは、Markdown文書を解析する際の基本的な文字と行の取り扱いを明確に定義しています。​これにより、さまざまなMarkdown実装間での一貫性が保たれ、予測可能な動作が保証されます。
