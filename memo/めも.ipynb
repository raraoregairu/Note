- 分類

css単位は大きく2つに分類可能

1. ほかの要素や画面サイズに影響を受けず常に固定値を持つ絶対値の単位(Absolute Units)

2. ほか要素の単位・大きさや画面サイズに応じて可変的な値となる相対値の単位(Relative Units)

- 絶対値の単位
    
    - cm<br>
    センチ(centi)は，国際単位系(SI)におけるSI接頭語の一つで，基礎となる単位の$10^{-2}$倍の量であることであるを示す．<br>
    センチは，ラテン語で「百」を意味するcentumに由来する．<br>
    
    - mm<br>
    ミリ(milli)は，国際単位系(SI)におけるSI接頭語の一つで，基礎となる単位の$10^{-3}$倍の量であることであるを示す．<br>
    ミリは，ラテン語で「千」を意味するmilleに由来する．<br>
    
    - in<br>
    ディスプレイのサイズを示す単位である．<br>
    １インチは，2.54cmに相当する．<br>
    また[ディスプレイの大きさ](#ディスプレイの大きさ)を参照<br>
    
    -pc<br>
    印刷業界における伝統的な長さの単位であり，主にタイポグラフィやレイアウトの設計に使用される．<br>
    １パイカは，4.233mm，または1/6インチに相当し，12ポイント（下に記述）で構成される．

    - pt<br>
    文字の大きさ（フォントサイズの単位）である．
    １ポイントは0.3528mm.
    
    - px<br>
    ピクセル(pixel)の略．1pixelはコンピュータで色を扱うための最小単位である．<br>
    1pxは1インチの1/96(約0.2645mm)，また1px = 3/4pt<br>
    
※dpiとppi

dpi(dots per inch)とは，おもにプリンタやイメージスキャナなどで使われる解像度の単位．<br>
幅１インチを何個の点（ドット）で表現できるかを表す値．
たとえば300dpiは紙面上の1インチあたり幅に300個，面積1平方インチあたりに9万個の微細な点をあらわす．

ディスプレイなどの表示装置では，解像度の単位として幅1インチ当たりの画素（ピクセル）数を表す「ppi」(pixels per inch)<br>
がもちいられることがあるが，表示装置ではドットもピクセルも同じため，dpiとppiも同義である．

$ppi = \sqrt{horizontal^2 + vertical^2}  / inch$

    - 画像の種類<br>

        画像のデータの構造（整理の仕方）には，大きく分けて
        「ピクセルの集まりとして記録する方法（ラスタ画像）」と，
        「図形の情報として記録する方法（ベクタ画像）」の2つがある．

        - ラスタ画像

            画像全体をピクセルの格子状の集まりとして捉え，各ピクセルに色の数値を記録する．
            例）100 $times$ 100ピクセルの画像→各１万点の色の数値で画像全体を表現
            解像度に依存し，拡大縮小で劣化する．
           

        - ベクタ画像

            図形や線を数式や座標情報（始点・終点，制御点）として記録する．
            例）円は中心の座標と半径で，直線は２点の座標で表現
            解像度に依存せず，拡大縮小しても劣化しない

        ※すべての画像は，ディスプレイ上では最終的にピクセル（ラスタ）として表示される．
        ただし，元のデータ構造がピクセルの集まり（ラスタ）なのか，図形の情報（ベクタ）なのかという違いが重要．

- 画像の構成

    - 画素

        画像はピクセル（picture element）と呼ばれる細かい四角形（正方形か長方形）が集めって構成されており，
        コンピュータで画像を扱う際に色情報（色調や諧調）を持つ最小単位である．<br>

    - 画素数

        画素数とは画像を構成するピクセルの総数である．

        画素数は画像の【横のピクセル数】$\times$【縦のピクセル数】で計算できる．

    - アスペクト比
    
        画像の縦と横の長さの比率
        
        
    - 解像度

        解像度とは、ある空間・領域・対象を，どれだけ小さな単位に分割して情報を記述できるか，またはその分割単位ごとの情報をどれだけ正確に記録・再現できるかを示す指標である．

        すなわち，解像度が高いとは，

        ・より高密度に情報が配置されている（空間あたりの情報量が多い）

        ・その情報が高い精度で記述されている（誤差や曖昧さが少ない）

        という2つの点で、情報の「詳細さ」や「正確さ」を両立している状態を意味する．

     - 解像度の種類
     
         - 絶対解像度
         
             情報そのものの細かさ，データとしての画素数．（情報）
         
              - 画像解像度

                 画像が内部的に保持しているピクセルの総数を示し，その画像がどれだけ細かい情報を持っているか，
                 すなわちどれだけ高密度かつ詳細に記録されているかを表す指標
         
             - ディスプレイ解像度

                 ディスプレイ上に表示可能な画素（ピクセル）の総数を示す指標．
                 通常は水平方向と垂直方向の画素数(例：1980 $\times$ 1080)で表され，
                 画面でどれだけ細かく画像や情報を表示できるかを決定する．
                 
             - 空間解像度
               
                 画像やデータがどれだけ細かい地理的・空間的な情報を表現できるかを示す指標．
                 具体的には１ピクセルが代表する実際の距離やどれだけ細かく地理的な情報を表現できるかを示す．
                
                 
         - 相対解像度
         
             単位面積あたりの画素の密度．
             情報をどれだけ精密に表示・出力できるか．（見え方）

             - ピクセル密度(PPI/DPI)

                 １インチあたりに配置されるピクセル（またはドット）数を示す指標．
                 ピクセル密度が高いほど，同じ面積内により多くの情報が詰め込まれるため，
                 画像や文字がより滑らかで精細に見える．

             - 印刷解像度(DPI)

                 印刷時に，１インチ当たりに打ち出されるインクのドット数を表す指標．
                 高いDPIは，より緻密で滑らかな印刷表現を可能とする．
         
             - 物理的解像度

                 画像のピクセル数と表示・印刷サイズとの組み合わせで決まる，実際の見た目の精細さの指標．
                 画像や画面がどれだけ高密度に実空間へ情報を投影しているかを表す．
                 
 　　 - 画像のサイズ

         画像を表示するために必要な物理的スペースを指す．
         ここでは，画像の実際の寸法(幅×高さ)やファイルサイズ(バイト数)を意味する場合がある．
         
        1. 物理的な寸法（例：4 cm $\times$ 3 cm）
        2. ファイルサイズ（例：5MBの画像ファイル）
        3. ピクセルサイズ（例：1980 $\times$ 1020ピクセル）
        
            - ファイルサイズ
            
                画像データを保存したときに記憶装置上で占める容量のことを指す．
                単位は「バイト(B)」
             
                 - ファイルサイズに影響する主な要素
                 
                     - 画素数（解像度）：ピクセルの総数が多いほど，記録すべきデータ量が増える
                     
                     - 色の情報（ビット深度）：１ピクセルにどれだけの色情報を持たせるか
                     
                     - 圧縮の有無：画像形式によっては「圧縮」によりファイルサイズが小さくなる．
                                   可逆圧縮と不可逆圧縮がある．
                                   
                     - 画像形式：同じ画像でも形式によってファイルサイズが大きく異なる．
                     
                     - メタデータ：撮影日時やカメラ情報，GPS情報など．
                     
- 画像の表示と処理

    画像は，内部的には色や明るさの情報を持つピクセルの集合として記録されている．
    しかし，そのピクセル情報がそのまま物理的に表示されるわけではない．
    画像データは表示装置（ディスプレイや印刷機など）で物理的に割り当てられ表示される．
    
    このとき，以下の変換・調整が行われる．
    
    - 画像のピクセル情報を出力装置の画素配置（相対解像度）に対応付ける．
    
    - 表示領域にあわせて，画像の拡大・縮小や補間処理を行う
    
    - 必要に応じて，色空間の変換やガンマ補正などの処理を挟む．
    
    この結果として，画像の大きさや見た目の滑らかさが出力装置によって異なることになる．

## 最小公倍数と最大公約数

最大公約数は２つ以上の整数に共通する約数（割り切れる数値）の中で一番大きな数値である．

- **ユークリッドの互除法**とは

    ２つの自然数$a, b$に対して，<br>
    $a = bq + r$である（$a$を$b$で割った商が$q$，余りが$r$）ならば，$a$と$b$の最大公約数は，$b$と$r$の最大公約数と同じである．

    このユークリッドの互除法を用いることで，２つの自然数の最大公約数を求めることができる．
    
    【例】
    
    3551と1961の最大公約数をユークリッド互除法で求める．
    
    $3551 \div 1961 = 1 \cdots 1590$ <br>
    $1961 \div 1590 = 1 \cdots 371$ <br>
    $1590 \div 371 = 4 \cdots 106$ <br>
    $371 \div 106 = 3 \cdots 53$ <br>
    $106 \div 53 = 2 \cdots 0$ 
    
    よって，3551と1961の最大公約数は **53**となる．
    
    【証明】

    条件より　$a = bq + r \cdots ①$ <br>
    移項すると　$r = a - bq \cdots ➁$

    $a$と$b$の最大公約数を $m$，$b$と$r$の最大公約数を$n$とする．

    $m$は$a$と$b$の最大公約数だから，

    $a = ma^{\prime}, b = mb^{\prime}$とすると，

    $r = ma^{\prime} - mb^{\prime}q = m(a^{\prime} - b^{\prime}q)$

    ➁より，$m$は$r$の約数といえる．

    よって，$m$は$b$と$r$の公約数といえる．

    $b$と$r$の最大公約数は$n$だから

    $m \leqq n \cdots③$

    一方，$n$は$b$と$r$の最大公約数だから，①より，$n$は$a$の約数といえる．

    $b = na^{\prime \prime}, r = nr^{\prime}$とすると，

    $a = na^{\prime \prime}q + nr^{\prime} = n(a^{\prime \prime}q + r^{\prime})$

    よって，$n$は$a$と$b$の公約数といえる．

    $a$と$b$の最大公約数は$m$だから，

    $m \geqq n \cdots④$

    ③，④より　$m = n$

    よって，「**割られる数と割る数の最大公約数は，割る数とあまりの最大約数と等しい**」．

    この操作を繰り返し行うと，余りは割る数より小さいので，この操作は有限回数で終わり，
    最後は必ず余りが０になり，最終的に最大公約数が求まるといえる．
