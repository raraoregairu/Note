- 分類

css単位は大きく2つに分類可能

1. ほかの要素や画面サイズに影響を受けず常に固定値を持つ絶対値の単位(Absolute Units)

2. ほか要素の単位・大きさや画面サイズに応じて可変的な値となる相対値の単位(Relative Units)

- 絶対値の単位
    
    - cm<br>
    センチ(centi)は，国際単位系(SI)におけるSI接頭語の一つで，基礎となる単位の$10^{-2}$倍の量であることであるを示す．<br>
    センチは，ラテン語で「百」を意味するcentumに由来する．<br>
    
    - mm<br>
    ミリ(milli)は，国際単位系(SI)におけるSI接頭語の一つで，基礎となる単位の$10^{-3}$倍の量であることであるを示す．<br>
    ミリは，ラテン語で「千」を意味するmilleに由来する．<br>
    
    - in<br>
    ディスプレイのサイズを示す単位である．<br>
    １インチは，2.54cmに相当する．<br>
    また[ディスプレイの大きさ](#ディスプレイの大きさ)を参照<br>
    
    -pc<br>
    印刷業界における伝統的な長さの単位であり，主にタイポグラフィやレイアウトの設計に使用される．<br>
    １パイカは，4.233mm，または1/6インチに相当し，12ポイント（下に記述）で構成される．

    - pt<br>
    文字の大きさ（フォントサイズの単位）である．
    １ポイントは0.3528mm.
    
    - px<br>
    ピクセル(pixel)の略．1pixelはコンピュータで色を扱うための最小単位である．<br>
    1pxは1インチの1/96(約0.2645mm)，また1px = 3/4pt<br>
    
※dpiとppi

dpi(dots per inch)とは，おもにプリンタやイメージスキャナなどで使われる解像度の単位．<br>
幅１インチを何個の点（ドット）で表現できるかを表す値．
たとえば300dpiは紙面上の1インチあたり幅に300個，面積1平方インチあたりに9万個の微細な点をあらわす．

ディスプレイなどの表示装置では，解像度の単位として幅1インチ当たりの画素（ピクセル）数を表す「ppi」(pixels per inch)<br>
がもちいられることがあるが，表示装置ではドットもピクセルも同じため，dpiとppiも同義である．

$ppi = \sqrt{horizontal^2 + vertical^2}  / inch$

    - 画像の種類<br>

        画像のデータの構造（整理の仕方）には，大きく分けて
        「ピクセルの集まりとして記録する方法（ラスタ画像）」と，
        「図形の情報として記録する方法（ベクタ画像）」の2つがある．

        - ラスタ画像

            画像全体をピクセルの格子状の集まりとして捉え，各ピクセルに色の数値を記録する．
            例）100 $times$ 100ピクセルの画像→各１万点の色の数値で画像全体を表現
            解像度に依存し，拡大縮小で劣化する．
           

        - ベクタ画像

            図形や線を数式や座標情報（始点・終点，制御点）として記録する．
            例）円は中心の座標と半径で，直線は２点の座標で表現
            解像度に依存せず，拡大縮小しても劣化しない

        ※すべての画像は，ディスプレイ上では最終的にピクセル（ラスタ）として表示される．
        ただし，元のデータ構造がピクセルの集まり（ラスタ）なのか，図形の情報（ベクタ）なのかという違いが重要．

- 画像の構成

    - 画素

        画像はピクセル（picture element）と呼ばれる細かい四角形（正方形か長方形）が集めって構成されており，
        コンピュータで画像を扱う際に色情報（色調や諧調）を持つ最小単位である．<br>

    - 画素数

        画素数とは画像を構成するピクセルの総数である．

        画素数は画像の【横のピクセル数】$\times$【縦のピクセル数】で計算できる．

    - アスペクト比
    
        画像の縦と横の長さの比率
        
        
    - 解像度

        解像度とは、ある空間・領域・対象を，どれだけ小さな単位に分割して情報を記述できるか，またはその分割単位ごとの情報をどれだけ正確に記録・再現できるかを示す指標である．

        すなわち，解像度が高いとは，

        ・より高密度に情報が配置されている（空間あたりの情報量が多い）

        ・その情報が高い精度で記述されている（誤差や曖昧さが少ない）

        という2つの点で、情報の「詳細さ」や「正確さ」を両立している状態を意味する．

     - 解像度の種類
     
         - 絶対解像度
         
             情報そのものの細かさ，データとしての画素数．（情報）
         
              - 画像解像度

                 画像が内部的に保持しているピクセルの総数を示し，その画像がどれだけ細かい情報を持っているか，
                 すなわちどれだけ高密度かつ詳細に記録されているかを表す指標
         
             - ディスプレイ解像度

                 ディスプレイ上に表示可能な画素（ピクセル）の総数を示す指標．
                 通常は水平方向と垂直方向の画素数(例：1980 $\times$ 1080)で表され，
                 画面でどれだけ細かく画像や情報を表示できるかを決定する．
                 
             - 空間解像度
               
                 画像やデータがどれだけ細かい地理的・空間的な情報を表現できるかを示す指標．
                 具体的には１ピクセルが代表する実際の距離やどれだけ細かく地理的な情報を表現できるかを示す．
                
                 
         - 相対解像度
         
             単位面積あたりの画素の密度．
             情報をどれだけ精密に表示・出力できるか．（見え方）

             - ピクセル密度(PPI/DPI)

                 １インチあたりに配置されるピクセル（またはドット）数を示す指標．
                 ピクセル密度が高いほど，同じ面積内により多くの情報が詰め込まれるため，
                 画像や文字がより滑らかで精細に見える．

             - 印刷解像度(DPI)

                 印刷時に，１インチ当たりに打ち出されるインクのドット数を表す指標．
                 高いDPIは，より緻密で滑らかな印刷表現を可能とする．
         
             - 物理的解像度

                 画像のピクセル数と表示・印刷サイズとの組み合わせで決まる，実際の見た目の精細さの指標．
                 画像や画面がどれだけ高密度に実空間へ情報を投影しているかを表す．
                 
 　　 - 画像のサイズ

         画像を表示するために必要な物理的スペースを指す．
         ここでは，画像の実際の寸法(幅×高さ)やファイルサイズ(バイト数)を意味する場合がある．
         
        1. 物理的な寸法（例：4 cm $\times$ 3 cm）
        2. ファイルサイズ（例：5MBの画像ファイル）
        3. ピクセルサイズ（例：1980 $\times$ 1020ピクセル）
        
            - ファイルサイズ
            
                画像データを保存したときに記憶装置上で占める容量のことを指す．
                単位は「バイト(B)」
             
                 - ファイルサイズに影響する主な要素
                 
                     - 画素数（解像度）：ピクセルの総数が多いほど，記録すべきデータ量が増える
                     
                     - 色の情報（ビット深度）：１ピクセルにどれだけの色情報を持たせるか
                     
                     - 圧縮の有無：画像形式によっては「圧縮」によりファイルサイズが小さくなる．
                                   可逆圧縮と不可逆圧縮がある．
                                   
                     - 画像形式：同じ画像でも形式によってファイルサイズが大きく異なる．
                     
                     - メタデータ：撮影日時やカメラ情報，GPS情報など．
                     
- 画像の表示と処理

    - 配列と座標系
    
        画像は，色や明るさなどの情報を，それぞれの位置に対応する画素単位で持っている．
        
       このような情報は，画像全体としては２次元の格子構造，すなわち行と列からなる配列（２次元配列）として記録される．
       
       画像という２次元配列において，「どの画素がどこにあるか」を決めるルール（座標系）が最初に決まっており，
       そのルールに従って配列の行番号(i)と列番号(j)が割り当てられている．
       
       画像の配列を表示領域に出力する際，画像の配列の変換や補間処理を行い，表示領域の各セルに適切な画素値を割り当てる必要がある．
       
        - 座標

            位置を数字で表す仕組み．<br>

            １次元なら数直線上の点の位置，２次元なら平面上の位置を，２つの数(列，行) であらわす．

       - 座標系

            座標で位置を表すための基準．

            - 原点の位置

                どの点を基準(0, 0)とするか．
                →画像やディスプレイでは左上，数学では左下や中心が多い．

            - x軸・y軸の向き

                横・縦方向に正の値がどちらに増えるか．
                →画像ではxが右，yは下．数学ではyは上に増えることが多い．x軸とy軸は直交しておる．

            - 単位の長さ（スケール）

                １単位(１ピクセル,1cm,1mなど）が実際にどれくらいの長さを表すか．
                →画像では「１画素」，地図では「１単位＝10m」など

            - 座標の値の形式

                - 連続座標系：実数(x, y)の値で表される．例）ベクトル描画や数学モデル


                - 離散座標系：位置は点ではなくセルとして扱う．
                       各セルはj列（横方向）とi行（縦方向）のインデックスで識別され,
                          すべて整数の組(j,i)で構成される．


                    例）ディスプレイや画像処理

                GISでは，「ベクタ形式（連続座標系）」と「ラスタ形式（離散座標系）」の量が使われる．

                ベクタは，点・線・面の幾何形状を連続空間上の座標（実数値）によって正確に表現する．

                一方，ラスタは各画素が固定サイズの地表範囲（例：10m $\times$ 10m）に対応することで，
                連続空間上の情報を離散的に近似して表現している．

                | 項目          | 連続座標系             | 離散座標系             |
                | ----------- | ----------------- | ----------------- |
                | 値の型         | 実数（float, double） | 整数（int）           |
                | 座標間の補間      | 可能                | 不可能（補間が必要なら補助処理）  |
                | 座標の意味       | 点の正確な位置           | セルのインデックス（番号）     |
                | 適した用途       | モデル化、解析、数学処理など    | 表示、保存、処理、画像処理全般など |
                | 格子構造への依存    | なし／弱い             | 強い                |
                | ピクセル（画素）の概念 | 基本的にない            | 基本的な単位            |

           - 座標の次元

               ２次元（画像，ディスプレイ）や３次元（3Dグラフィックスなど）

           - 回転・反転の有無

               画像処理では上下反転や回転された座標系を使うことがある．


       - 画像・ディスプレイの座標系
           - 原点：左上が(0, 0)
           - x軸：右に行くほど増加
           - y軸：下に行くほど増加
           - 座標の値の形式：離散座標系

        - 配列
        
            画像は，各画素の色情報や明るさの値を格納した２次元配列として内部的に表現される．
            
            この配列はがぞの空間的な並びに対応するように構成されており，各要素（セル）は画像上の位置を表す行(i)と列(j)
            のインデックスを持つ．
            
            行(i)は縦方向，列(j)は横方向に増加する．
            
            これらのインデックスは前述の座標系（左上を原点として，x座標・y座標が増加する方向が右・下である格子状の座標系）　　
            に基づいて定義される．
            
            つまり，配列のインデックスｊは画像の横方向(x軸)に対応し，iは縦方向(y軸)に対応する．
            
            例えば，縦200行・横300列の画像の場合，画像の座標系における有効範囲はx = 0～299（横方向），y = 0～199（縦方向），
            配列のインデックスはi = 0～199，j = 0～299となる．
            
            
            ```
            配列（i行, j列）:           → j（横方向: x軸）
                              0    1    2    3
                            +----+----+----+----+
                    i = 0   |100 |110 |120 |130 |
                            +----+----+----+----+
                    i = 1   |140 |150 |160 |170 |
                            +----+----+----+----+
                    i = 2   |180 |190 |200 |210 |
                            +----+----+----+----+
            ↓
            i（縦方向: y軸）
            ```

            例）縦３行×横４列の画像
            
            例えば(i, j) = (1, 2) の要素 160 は、画像上の座標系で見ると (x, y) = (2, 1) にあるピクセルの値である．
            
            そのため，配列の中の要素(i,j)は」，画像上の特定の位置
            🔍 行×列？列×行？―用途で変わる画像の表記順
            - 画像データの構造：プログラムや内部処理では「行 × 列（[y][x]）」で扱うのが一般的である．<br>
                 → 上から順に行，横に並ぶのが列。

            - 解像度やスペック表記：人の視覚やディスプレイの見た目に合わせて「横 × 縦（列 × 行）」の順で表記される．<br>
                 → 例：1920 × 1080（横1920ピクセル、縦1080ピクセル）

            - 混乱しやすいポイント：どちらも正しいけれど「文脈に応じた表記」で使い分ける必要がある．
  

 - 行列と画像処理について
 
     画像処理において，行列（Matrix）は，非常に重要な役割を果たす．
     
     行列は数や式を縦横に並べた矩形の配列であり，数学や工学の分野において広く使用される．
     
     画像もまた，二次元の配列であるため，行列と画像処理は密接に関連している．<br><br>
 
    - 行列とは
    
        行列とは下記の配列のように数値や式を縦と横に規則的に整えて並べたものである．
    
    	\begin{array}{cc}
           1 & 2 \\
           3 & 4
        \end{array}
    
        水平方向のまとまりを行という．０行目の配列は[１，２]，１行目の配列は[３，４]となる．

        \begin{array}{c@{\hskip 5px}c}
        1 & 2 \\ \hline
        3 & 4
        \end{array}

        垂直方向のまとまりを列という．０列目の配列は[１，３]，１列目の配列は[２，４]となる．
        
        \begin{array}{c|c}
        1 & 2 \\ 
        3 & 4
        \end{array}        
        
        行列に並んだ数や文字のひとつひとつを行列の成分という．１，２，３，４はそれぞれ行列の成分である．
        また，$i$ 行 $j$ 列にある成分を$(i, j)$成分という．
        一般に，$m$個の行と$n$個の列からなる行列を$m$行$n$列の行列または$m \times n$行列という．<br>
        また，この行列の型は$m \times n$であるとも，この行列は$m \times n$であるともいえる．
        コンピュータで行列を扱う場合は行や列の番号は０から始まるのが一般的である．<br>
        このため，$m$行$n$列の配列では，行は $0$ から $m - 1$，列は $0$ から $n - 1$ までとなる．<br>
       
        
        \begin{bmatrix} 
          a_{0,0} & \dots & a_{0, j} & \dots & a_{0, n- 1} \\
          \vdots &       & \vdots &       & \vdots \\
          a_{i, 0} & \dots & a_{i, j} & \dots & a_{i, n-1} \\
          \vdots &       & \vdots &       & \vdots \\
          a_{m-1, 0} & \dots & a_{m - 1, j} & \dots & a_{m - 1, n - 1}
        \end{bmatrix} 

        ```
        配列（i行, j列）a(i,j):   
                            → j（横方向: x軸）

                          0       1       2       3
                    +--------+--------+--------+--------+
            i = 0   | a(0,0) | a(0,1) | a(0,2) | a(0,3) |
                    +--------+--------+--------+--------+
            i = 1   | a(1,0) | a(1,1) | a(1,2) | a(1,3) |
                    +--------+--------+--------+--------+
            i = 2   | a(2,0) | a(2,1) | a(2,2) | a(2,3) |
                    +--------+--------+--------+--------+
        ↓
        i（縦方向: y軸）
        ```
   -  行列の型
   
         - 正方行列
         
             行と列の個数が等しいければ，全体の形としては正方形になる．
             特に$n$行$n$列の行列を$n$次正方行列という．
             
         - 数ベクトル
         
             以下の行列の総称．
         
             - 行ベクトル

                 行の個数が１である行列．
                 成分の個数が$n$個である行ベクトルを$n$次行ベクトルという．

             - 列ベクトル

                 列の個数が１である行列．
                 成分の個数が$n$である列ベクトルを$n$次列ベクトルという．
                 
    - 行列の定義と記法（一般形）

$$C = A \pm B = [a_{i,j} \pm b_{i,j}]$$

- 行列の性質

    - 行列の加法における演算の性質
    
        - 単位元
        
            ある演算において，ほかの要素と組み合わせてもその要素の値を変えない特別な要素のこと．
            
            任意の行列$A$に対して
            
            $$A + O = [a_{i,j} + 0] = [a_{i,j}] = A$$

            また，
        
            $$O + A = [0 + a_{i,j}] = [a_{i,j}] = A$$

            となるから零行列$O$は，行列の加法における単位元である．
        
        - 逆元
        
            要素に対する演算の結果が単位元になる要素のこと．
            
            任意の行列$A$に対して
        
            $$A + X = O $$

            もしくは，
        
            $$X + A = O $$

            を満たすような行列Xを-Aと表す．

            行列$A=[a_{i,j}]$に対して

            $$(-1)A = [-a_{i,j}] $$

            このとき，

            $$A + (-1)A = [a_{i,j}] + [-a_{i,j}] = [a_{i,j} + (-a_{i,j})] = [0] = O$$

            よって行列$A$に対して，定数倍$(-1)A$を加えると零行列になるため，加法逆元$-A$に等しい．

            よって$-A = (-1)A$が成り立つ．

            そして減法は

            $$A - B = A + (-B)$$

            とすることで加法に置き換えられる．

        - 交換律
        
            要素の順番を入れ替えても演算の結果が変わらない．

            行列$A, B$に対して
        
            $$A + B = B + A$$

        - 結合律
        
            複数の要素の演算において，演算の順番を変えても結果が同じである．

            行列$A, B, C$に対して
        
            $$(A + B) + C= A + (B + C)$$

    - 行列の定数倍における演算の性質
    
        - 1倍の行列（単位元）

           任意の行列$A$に対して，$1$をかけると，元の行列と変わらない．<br>

            $$1 \cdot A = A$$

        - 零元をもとめる
        
            零元とはある演算においてほかの要素と組み合わせてもその要素の値を変えない特別な要素であり，<br>
            行列の定数倍では以下の演算の結果で生み出される．
            
            - 0倍の行列

                任意の行列$A$に対して，$0$をかけると，必ず零行列$O$になる．

                $$0 \cdot A = O$$

            - 零行列の定数倍

                任意の定数 $k$ に対して、零行列を $k$ 倍すると、結果は必ず零行列になる．

                $$k \cdot O = O$$

        - 結合律
        
            演算の順番を変えても結果が同じである．

            行列$A$に対して
        
            $$k(lA) = (kl)A$$
            
        - 分配律
            
            各項に対して個別に演算しても，先に括弧内の演算を計算してから全体を計算しても結果が同じである．
            
            行列$A$に対して
        
            $$(k + l)A = kA + lA$$
            
            また，行列$A, B$に対して
            
            
            $$k(A + B) = kA + kB$$

    行ベクトルに列ベクトルをかける計算を$(i, j)$成分とする行列$C= AB$(サイズ$m \times p$) を$A,B$の積という．

    - 単位元

         まず，行ベクトル×列ベクトル（２次元としてみた行列）から考える

         任意の$n$個の成分をもつ行ベクトル$A$ および列ベクトル$B$がある．
         
         その積は，
         
    $$
        \begin{bmatrix} 
          \scriptstyle a_0 & \scriptstyle a_{1} & \dots& \scriptstyle a_{n-1} \\
        \end{bmatrix}
        \ \cdot\ 
        \begin{bmatrix} 
          \scriptstyle b_{0}  \\
          \scriptstyle b_{1}  \\
          \vdots  \\
          \scriptstyle b_{n-1}
        \end{bmatrix}
        \ =\ 
        \begin{matrix} 
          \scriptstyle a_{0} b_{0}  & + & \scriptstyle a_{1} b_{1} & \dots &  + &\scriptstyle a_{n-1} b_{n-1} \\
        \end{matrix}
    $$

    単位元は演算の結果によって元の成分と変わらないものである．
    ここでは，行ベクトル$A$の中のある成分だけ抽出する操作を考える．
        
    たとえば，$A$の0番目の成分$a_0$を取り出したいなら

    $$
        \begin{bmatrix} 
          \scriptstyle a_{0} & \scriptstyle a_{1} & \dots& \scriptstyle a_{n-1} \\
        \end{bmatrix}
        \ \cdot\ 
        \begin{bmatrix} 
          \scriptstyle 1  \\
          \scriptstyle 0  \\
          \vdots  \\
          \scriptstyle 0
        \end{bmatrix}
        \ =\ 
        \begin{matrix} 
          \scriptstyle a_{0} \times 1  & + & \scriptstyle a_{1} \times 0 & \dots &  + &\scriptstyle a_{n-1} \cdot 0 \\
        \end{matrix}
        \ = a_{0}\
    $$
    
    同様に，$a_1$を取り出すには
    $$
        \begin{bmatrix} 
          \scriptstyle a_{0} & \scriptstyle a_{1} & \dots& \scriptstyle a_{n-1} \\
        \end{bmatrix}
        \ \cdot\ 
        \begin{bmatrix} 
          \scriptstyle 0  \\
          \scriptstyle 1  \\
          \vdots  \\
          \scriptstyle 0
        \end{bmatrix}
        \ =\ 
        \begin{matrix} 
          \scriptstyle a_{0} \times 0  & + & \scriptstyle a_{1} \times 1 & \dots &  + &\scriptstyle a_{n-1} \cdot 0 \\
        \end{matrix}
        \ = a_{1}\
    $$
   つまり，$a_i$を取り出すには，$i$成分が１でほかがすべて０である列ベクトルを用いる．
   
    $$
        \begin{bmatrix} 
          \scriptstyle a_{0} & \scriptstyle a_{1} & \dots& \scriptstyle a_{n-1} \\
        \end{bmatrix}
        \ \cdot\ 
        \begin{bmatrix} 
          \scriptstyle 0  \\
          \vdots  \\
          \scriptstyle 1(i成分)  \\
          \vdots  \\
          \scriptstyle 0
        \end{bmatrix}
        \ =\ 
        \begin{matrix} 
          \scriptstyle a_{0} \times 0  & + &  \dots & + &\scriptstyle a_{i} \times 1 &+ &\dots & + &\scriptstyle a_{n-1} \cdot 0 \\
        \end{matrix}
        \ = a_{0, i}\
    $$
    
    このような「１つの位置にだけ１があり，ほかは０である列」は，行ベクトルの中にある位置（列番号）を特定するための基本的な構成単位として利用できる．
    
    このような列を$e_k$と記すことにすると：
    
    $$
        \ e_{k} =\ 
        \begin{bmatrix} 
          \scriptstyle 0  \\
          \vdots  \\
          \scriptstyle 1(k成分)  \\
          \vdots  \\
          \scriptstyle 0
        \end{bmatrix}
    $$
    
   　この$e_k$を使うと，行ベクトル$A$の$k$成分は$A \cdot e_k$によって取り出せる．  
    　
   行ベクトルは各成分から左から右の順番で並べられているため，$e_k$を左から右に並べて行列$I$を作る．
    
    $$I = \begin{bmatrix} e_0 & e_1 & \cdots & e_{n-1} \end{bmatrix}$$
    
    この$I$によって
    $$A \cdot I = A$$
    となり，元の行ベクトルを丸ごと再現できる．
    ここでは行列$A$の各列を保持するような操作が行われている．
    
    これを単位行列という．
    
    この操作は行列の積の一般式から行列$A$に対してもなりたつ．

    画像の中の位置(x, y)を，画面上の位置(x',y')に表示するか対応づける．

    単純に画像と表示領域が同じサイズなら，マッピングはそのまま(１対１）．

    サイズが違う場合は画像の拡大や縮小（スケーリング）が必要である．

    スケーリングでは，元の画像サイズと表示領域のサイズを元に，拡大・縮小の比率（スケーリング係数）を使って<br>
    ピクセルの位置を変換する．

    ①スケーリングとマッピングの基本と定義

    ◇パラメータの定義
    | パラメータ           | 意味            |
    | ---------------- | ------------- |
    | $W_{\text{in}}$  | 元画像の横幅（ピクセル数） |
    | $H_{\text{in}}$  | 元画像の高さ（ピクセル数） |
    | $W_{\text{out}}$ | 表示先の横幅（ピクセル数） |
    | $H_{\text{out}}$ | 表示先の高さ（ピクセル数） |
    | $S_x$            | 横方向のスケーリング係数  |
    | $S_y$            | 縦方向のスケーリング係数  |

    ◇スケーリング係数

    $$S_x = \cfrac{W_{\text{out}}}{W_{\text{in}}}, S_y = \cfrac{H_{\text{out}}}{H_{\text{in}}}$$

    ◇ピクセル座標の変換式

    元画像の座標$(x, y)$を表示領域の座標$(x', y')$に変換するには：

    $$x' = x \cdot S_x, y' = y \cdot S_y$$

    ➁計算例

    ✅ 条件

     画像：400px $\times$ 300px
     表示領域：800px $\times$ 600px

     ✅ スケーリング係数の計算

     $$S_x = \cfrac{800}{400} = 2.0,S_y = \cfrac{600}{300} = 2.0$$

     ✅ 画像のピクセル$(x, y)=(100, 50)$の場合

    $$x' = 100 \cdot 2.0 = 200, y' = 50 \cdot 2.0 = 100$$

    →ディスプレイ上の表示位置は$(200, 100)$

    ③補足：スケーリング係数が整数でない場合

    例えば，表示領域が600px $\times$ 450pxの場合

    $$S_x = \cfrac{600}{400} = 1.5,S_y = \cfrac{450}{300} = 1.5$$

    →画像のピクセル$(x, y) = (101, 51)$の変換後の位置

    $$x' = 101 \cdot 1.5 = 151.5, y' = 51 \cdot 1.5 = 76.5$$

    このように整数座標でなくなる場合，補間処理が必要になる．

- 拡大・縮小と補間処理

    マッピングにより画像のサイズと表示領域の催事が異なる場合画像の拡大縮小をする必要がある．しかし

## 最小公倍数と最大公約数

最大公約数は２つ以上の整数に共通する約数（割り切れる数値）の中で一番大きな数値である．

- **ユークリッドの互除法**とは

    ２つの自然数$a, b$に対して，<br>
    $a = bq + r$である（$a$を$b$で割った商が$q$，余りが$r$）ならば，$a$と$b$の最大公約数は，$b$と$r$の最大公約数と同じである．

    このユークリッドの互除法を用いることで，２つの自然数の最大公約数を求めることができる．
    
    【例】
    
    3551と1961の最大公約数をユークリッド互除法で求める．
    
    $3551 \div 1961 = 1 \cdots 1590$ <br>
    $1961 \div 1590 = 1 \cdots 371$ <br>
    $1590 \div 371 = 4 \cdots 106$ <br>
    $371 \div 106 = 3 \cdots 53$ <br>
    $106 \div 53 = 2 \cdots 0$ 
    
    よって，3551と1961の最大公約数は **53**となる．
    
    【証明】

    条件より　$a = bq + r \cdots ①$ <br>
    移項すると　$r = a - bq \cdots ➁$

    $a$と$b$の最大公約数を $m$，$b$と$r$の最大公約数を$n$とする．

    $m$は$a$と$b$の最大公約数だから，

    $a = ma^{\prime}, b = mb^{\prime}$とすると，

    $r = ma^{\prime} - mb^{\prime}q = m(a^{\prime} - b^{\prime}q)$

    ➁より，$m$は$r$の約数といえる．

    よって，$m$は$b$と$r$の公約数といえる．

    $b$と$r$の最大公約数は$n$だから

    $m \leqq n \cdots③$

    一方，$n$は$b$と$r$の最大公約数だから，①より，$n$は$a$の約数といえる．

    $b = na^{\prime \prime}, r = nr^{\prime}$とすると，

    $a = na^{\prime \prime}q + nr^{\prime} = n(a^{\prime \prime}q + r^{\prime})$

    よって，$n$は$a$と$b$の公約数といえる．

    $a$と$b$の最大公約数は$m$だから，

    $m \geqq n \cdots④$

    ③，④より　$m = n$

    よって，「**割られる数と割る数の最大公約数は，割る数とあまりの最大約数と等しい**」．

    この操作を繰り返し行うと，余りは割る数より小さいので，この操作は有限回数で終わり，
    最後は必ず余りが０になり，最終的に最大公約数が求まるといえる．

線分の内分の証明（一次元）


$m: n= x - x_0 : x_1- x$

$n(x - x_0) = m(x_1 - x)$

$(m+n)x = nx_0 + mx_1$

$x = \cfrac{nx_0 + mx_1}{m+n}$


x方向もしくはy方向の距離を比率ととらえて値（例えば高さz）を内分で求める！（二次元）
$z = \cfrac{(x_1 - x)z_0 + (x-x_0)z_1}{x_1-x_0}$

$z = \cfrac{x_1 - x}{x_1-x_0}z_0 + \cfrac{x-x_0}{x_1-x_0}z_1$

三次元の場合$y_0$を固定した場合と$y_1$を固定した場合のzを先に求める

①$y_0$でのz

$z_{*0} = \cfrac{x_1 - x}{x_1-x_0}z_{00} + \cfrac{x-x_0}{x_1-x_0}z_{10}$

➁$y_1$でのz

$z_{*1} = \cfrac{x_1 - x}{x_1-x_0}z_{01} + \cfrac{x-x_0}{x_1-x_0}z_{11}$

①,➁を使ってyの比率を使って最終的なzを求める

$z_{xy} = \cfrac{y_1 - y}{y_1-y_0}z_{*0} + \cfrac{y-y_0}{y_1-y_0}z_{*1}$

代入して

$z_{xy} = \cfrac{y_1 - y}{y_1-y_0} \left( \cfrac{x_1 - x}{x_1-x_0}z_{00} + \cfrac{x-x_0}{x_1-x_0}z_{10} \right) + \cfrac{y-y_0}{y_1-y_0}\left(\cfrac{x_1 - x}{x_1-x_0}z_{01} + \cfrac{x-x_0}{x_1-x_0}z_{11}\right)$

$x_1 - x_0$でくくって整理

$z_{xy} = \cfrac{1}{(y_1-y_0)(x_1-x_0)} \left( (x_1 - x)(y_1 - y)z_{00} + (x - x_0)(y_1 - y)z_{10}+ (x_1 - x)(y-y_0)z_{01} + (x-x_0)(y-y_0)z_{11}\right)$

単位格子で考えると$x_1 - x_0 = 1, y_1 - y_0=1$であり，$dx = x - x_0, dy = y - y_0$とすると

$
    x = x_0 + dx \\
    x = x_1 - 1 + dx \\
    y = y_0 + dy \\
    y = y_1 - 1 + dy
$

よって整理すると

$z_{xy} = z_{00} (1 - dx)(1 - dy) + z_{10}dx(1 - dy) + z_{01}(1 - dx)dy + z_{11}dxdy$
