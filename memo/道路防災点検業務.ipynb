# 道路防災点検業務

## 道路防災点検

豪雨・豪雪等による災害を防止するため，道路法面の安定性等について詳細な点検を行い，さらにその結果を今後の道路防災対策に反映していくためのものである．

点検後の対応策等を記した「防災カルテ」，「箇所別記録表」の作成や点検結果のデータベース化をする．

## 防災点検の流れ

## 第１絞込み

「点検対象区間」を選定することを目的として絞り込み（第１絞込み）を実施する．

第１絞込みは，管理対象道路の防災レベルを概括的に把握して，安定度調査の候補箇所を含む区間（点検対象区間）を選定する．

具体的には次の①～③のいずれかに該当する区間を選定する．

　①平成８年度点検における要対策箇所のうち対策未了箇所を含む一連の区間

　➁平成８年度点検以降に道路災害の発生した箇所及び災害の兆候が認められる箇所を含む一連の区間

　③その他，道路管理者が防災管理上必要と認める箇所を含む一連の区間

なお，点検対象区間の選定にあたっては，過去の災害履歴，点検履歴などに関する資料を参考にするほか，
必要に応じて道路防災ドクターや専門の点検技術者等の意見を聴取する．

## 第２絞込み

選定された点検対象区間を対象として，危険要因を抽出することにより，安定度調査を実施する必要がある箇所を絞り込む(第２絞込み)．

第２絞込みは，「机上調査」(第２絞込み①)と「現地確認」(第２絞込み②)の２つからなる．

このうち「机上調査」では，「地域特性の把握」及び「災害要因の判読」を実施する．

「現場確認」は「安定度調査候補箇所」について「机上調査」により判読した災害要因が現地においても見られるかどうかを確認する．

## 安定度調査の実施

安定度調査では，点検箇所の現地を観察し，箇所別記録表と安定度調査票を作成する．
箇所別記録表は，点検箇所の点検対象項目ごとに作成する．

点検対象項目

 - A:落石・崩壊
 - B:岩盤崩壊
 - C:地すべり
 - D:雪崩
 - E:土石流
 - F:盛土
 - G:擁壁
 - H:橋梁基礎の洗堀
 - I:地吹雪
 - J:その他（越波など）

擁壁：大きな高低差を地面に設ける際に，その崖となる土の側面が崩れ落ちるのを防ぐために設計・構築される経机上の建造物

洗堀：激しい川の流れや波浪などにより，堤防の表法面の土が削られる状態

地吹雪：地上から舞い上げられた雪が，激しい風で空中を乱れ飛ぶ状態

箇所別記録表には以下を記載する．

 - 管理者の名称
 - 路線名
 - 道路種別
 - 所在地などの基本的な情報
 - 点検箇所のスケッチ
 - 被災履歴
 - 安定度調査結果（総合評価）
 - 想定対策工　など
 
また，スケッチの中に斜面や構造物の状況の概要，安定度調査を実施する範囲（部分）ごとに部分番号を記載する．

安定度調査は，点検箇所の点検対象項目に従って，該当する調査票を用いる．調査票に沿って評価点数を求めたうえで総合評価を実施する．箇所別記録表で斜面の部分が複数示されている場合には部分ごとに作成する．

総合評価は，安定度調査を実施した箇所が，以下の３項目のいずれに該当するかを示す．

 - 対策が必要とされる（要対策）
 - 防災カルテを作成し対応する（カルテ対応）
 - 特に新たな対応を必要としない（対応不要）

## 安定度調査※変状例

**落石**

岩盤の不連続面（岩盤中に発達する節理，片理，層理等の割れ目）が拡大して，岩塊や礫が剥離したり，表層堆積物，火山噴出物，固結度の低い砂礫層の中の岩塊，礫が表面に浮き出して斜面より落下する現象

**浮石**

岩盤は自然の風雨に触れた部分から岩盤の割れ目に沿って雨水や粘土が流入していく風化作用を受ける．

この作用が長い時間を経て，割れ目が徐々に開いていき，基盤から遊離・脱落しそうに不安定状態で残った岩石である．

**転石**

基盤の岩石から遊離・脱落した岩石が，流水などに押し流されたり，山の斜面を転がり落ちたり，崖下などに転落している岩石である．

**崩壊跡**

斜面において，自然作用（雨・風等）および人工作用（斜面掘削・伐採等）を受けて，自然および人口斜面を安定した状態で保持できないとき，斜面が崩れ落ちてしまう（崩落）．この崩れ落ちた状態の斜面である．

**滑落崖**

斜面において，基盤となる岩盤が自然作用を受けて岩盤の割れ目沿いに滑り落ちてしまう．
この滑り落ちた状態が険しく切り立った崖状のところである．

**亀裂開口**

露岩および対策施設が何らかの原因にて亀裂が入っている．
※今後の変状を確認するため，鋲を打ち，鋲間を計測する．

## 取りまとめ調書

**箇所別記録表**

各点検箇所について，管理機関名，施設管理番号，路線名，位置情報等を記入し，必要なスケッチ等を行う．

また，安定度調査票の項目にない事項に関しては，特記事項の所見の欄に具体的に記述する．

**安定度調査票**

点検対象項目ごとの「安定度調査票」の記入は，該当するチェック項目に〇印を付け，評点等の数字を記入したうえで[要因],[対策工],[履歴]等の合計の欄を埋めるものとする．

さらに，[総合評価]として，現地状況を総合的に勘案して，対策の必要性等を判断し，該当欄に〇印を付す．

ただし，現地で既設対策工の有効性の判断ができない場合は，設計計算結果から，総合評価の判断をしてもよい．

また，「安定度調査票」では，状況を十分に記録することができない詳細な現地状況については，「箇所別記録表」や「防災カルテ」に記録する．

総合評価は次に示す３段階で評価する．

 - 対策が必要とされる：災害に至る可能性のある要因が，明らかに認められる箇所
 - 防災カルテを作成し対応する：将来的には対策が必要となる場合が想定されるものの，当面「防災カルテ」による監視等で管理していく箇所
 - 特に新たな対応を必要としない：災害の要因となるものが発見されず，特に新たな対応を必要としない箇所．

**被災履歴記録表**

資料調査により平成８年度点検以降の災害記録がある箇所では，その収集資料より転写や複写により被災履歴記録表を作成する．

なお，当記録表に記載しきれない図表や事項については，別紙に表してもよい．

 - 平面図・断面図・スケッチ/現況写真・対策工概要図
 - 特記事項（発生年月日，規模，誘因，被害，通行止実績，対策工施工年度，対策工種，対策工費，その他）

**防災カルテ**

防災カルテは，防災カルテ様式A，防災カルテB，防災カルテC，着目すべき変状のチェックリストから構成される．
